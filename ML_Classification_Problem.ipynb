```python
#convert these char to numbers like rbc,pc,pcc,ba,htn,dm,cad,appet,pe,ane coobject_columns

#Handling Encoding part of x 
#label encoder and one hot encoder in ML

def classify_features(x):
    categorical_features=[]
    non_categorical_features=[]
    discreate_features=[]
    continuous_features=[]
    for column in x.columns:
        if x[column].dtype == 'str':
            if x[column].nunique()<3:
                categorical_features.append(column)
            else:
                non_categorical_features.append(column)
        elif x[column].dtype in ['float64','int64']:
            if x[column].nunique() < 100:
                discreate_features.append(column)
            else:
                continuous_features.append(column)
    return categorical_features, non_categorical_features, discreate_features, continuous_features
```

```python
categorical_features, non_categorical_features, discreate_features, continuous_features = classify_features(x)
```
